```python
%%writefile garden.py
import spacy

nlp = spacy.load("en_core_web_sm")

gardenpathSentences =[
    "The old man the boat.",
    "The horse raced past the barn fell.",
    "Mary gave the child a Band-Aid.",
    "That Jill is never here hurts.",
    "The cotton clothing is made of grows in Mississippi."
]

for sentence in gardenpathSentences:
    doc = nlp(sentence)

    print("\nSentence:", sentence)

    print("Tokens:")
    for token in doc:
        print(token.text)

    print("Named Entities:")
    for ent in doc.ents:
        print(ent.text, ent.label_)

print(spacy.explain("PERSON"))
print(spacy.explain("GPE"))
print(spacy.explain("ORG"))
print(spacy.explain("NORP"))

# Entity 1: Jill was identified as PERSON.
# PERSON means people, including fictional characters.
# This makes sense because Jill is a person's name.

# Entity 2: Mississippi was identified as GPE.
# GPE means countries, cities and states.
# This makes sense because Mississippi is US state.
```
